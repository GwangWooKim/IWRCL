```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from eval import cal_ks_wd
```

```python
# Load data
u = np.load("../data/exp_u.npy")
generated = np.load("../save/exp/IWRCL/M100/seed42/generated.npy")

u = u.ravel()
generated = generated.ravel()

# Evaluation
ks, wd = cal_ks_wd(generated, u, "exp")
print(f"KS: {ks:.4f}")
print(f"WD: {wd:.4f}")

# Plot
plt.figure(figsize=(6, 4))

sns.kdeplot(u, label="True", linewidth=2)
sns.kdeplot(generated, label="IWRCL", linewidth=2)

plt.ylabel("Density")
plt.title(f"Exp | KS = {ks:.3f}, WD = {wd:.3f}")
plt.legend(frameon=False)

sns.despine()
plt.tight_layout()
plt.show()
```
